```python
# @title
class Hewan:
  def __init__(self,nama):
    self.nama = nama

  def makan(self):
    print(self.nama,"sedang makan.")

class Burung(Hewan):
  def terbang(self):
    print(self.nama,"sedang terbang.")

class Serangga(Hewan):
  def makan(self):
    print(self.nama,"sedang makan sedikit.")


manuk = Burung("Dadali")
manuk.terbang()
Hewan("Sapi").makan()
Serangga("Ulat").makan()
```

```python
from abc import ABC

class Hero:
    def __init__(self,name,hp,power):
        self.name = name
        self.hp = hp
        self.atkpower = power
    def info(self):
        print(f"""
Hero Name: {self.name}
Health: {self.hp}
Attack Power: {self.atkpower}
""")
```

```python
def main():
  h1 = Hero("One",200,1)
  h2 = Hero("Two",300,100)

  h1.info()
  h2.info()

if __name__ == "__main__":
  main()
```
